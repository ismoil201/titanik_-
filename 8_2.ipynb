## 배열의 형태 변경  (중요!!!)

### 형태 변경: reshape(), flatten()
### 통합과 분할:  vstack(), hstack() row_stack(), column_stack(), concatenate(), 

```python
### reshape

m1=np.arange(6)
print(m1)
print(m1.shape)

m2=m1.reshape(6, 1)
print(m2, m2.shape)
```

```python
m1=np.arange(6)
print(m1, m1.shape)

m2=m1.reshape(-1, 2) # 행 수는 자동, 2열 생성
print(m2, m2.shape)

m2=m1.reshape(-1, 1) # 행 수는 자동, 1열 생성
print(m2, m2.shape)

m2=m1.reshape(2, -1) # 행 수는 자동, 1열 생성
print(m2, m2.shape)
```

```python
### 다차원을 1차원으로 flatten
a=np.arange(12).reshape(3, 4)
print(a, '\n')

f=a.flatten()
print(f, f.shape, '\n')

b = a.reshape(-1, 12)
print(b, b.shape)
```

```python
### 배열의 방향 변경하기
#print(a, '\n')



##  중 요 ##
# slicing 규칙:  [start, stop, step].
# slicing 은 loop 를 시작 하기 전에 start, stop, step 의 index 값을 정한다.
# step:  지정되어 있지 않으면 +1 적용
#
# 갯수가 n 
# 1) step > 0 일때
#    start:
#       None:  0
#       음수:    n + start
#    stop:
#       None:  n 
#       음수:   n + start
# 2) step < 0 일때
#    start:  
#       None:   n - 1
#       음수:   n + start
#    stop:
#       None:  -n + -1
#       음수:  n + start
####################


#print(a[::-1])   # => a[2:-4:-1]

# print(a[0:2])
# print(a[0:2, :])    # a[0:2, :] == a[0:2]

print(a[:-1])     # => a[0:2:1]
print(a[0:2:1])

print(a[1:-1:-1])  #  => a[1:2:-1]
```

```python
# 열의 순서를 변경
print(a[:, ::-1])
print()

print(a[::-1, ::-1])
print()

print(a[2:-2:-1])
```

```python
print(a[2:0:-1])

print(a[1:-1:-1])

print(a[1:-1:1])
```

```python
t=a.transpose()
print(a, a.shape) # a 배열은 3행 4열
print(t, t.shape) # t 배열은 a 배열을 4행 3열로 변경한 것
```

```python
### 행령에 scalar 계산
a=np.arange(9).reshape(3, 3)
b=a*2

print(a)
print()

print(b)
```

## 배열 통합(join) 함수
#  concatenate, stack, vstack, hstack

```python
r=np.vstack((a, b))
print(r, r.shape)
```

```python
x=np.concatenate((a, b), axis=0)
print(x, x.shape)

y=np.concatenate((a, b), axis=1)
print(y, y.shape)
```

```python
r=np.hstack((a, b))
print(r, r.shape)
```

```python
c=np.column_stack((a, b))
print(c, c.shape)
```

```python
a1 = np.array([0,1,2])
a2 = np.array([10,11,12])
print(a1.shape)

h1 = np.hstack((a1,a2))
print(h1, h1.shape)

# column_stack:  hstack 과 비슷하지만 1차원 행렬을 합칠때는 각 배열을 열로 취급해서 2차원 배열로 만듬
h2 = np.column_stack((a1,a2))
print(h2, h2.shape)
```
