```python
###  numpy 배열

np.arange(12)
```

```python
### reshape

d=np.arange(12).reshape(3,4)
print(d,d.shape)
```

```python
###  ndim  (number of dimensions)

## scalar:   ndim = 0        ex) np.array(10)
## vector:   ndim = 1        ex) np.array([0,1,2,3,4])
## matrix:   ndim = 2        ex) np.array([[1,2], [3,4], [5,6]])
## tensor:   ndim >= 3       ex) np.array([ [[1,2], [3,4]], [[5,6], [7,8]] ])

d0 = np.array(10)
d1 = np.array([0,1,2,3,4])
d2 = np.array([[1,2], [3,4], [5,6]])
d3 = np.array([ [[1,2], [3,4]], [[5,6], [7,8]] ])

print(d0, ' ndim: ', d0.ndim, '\n') 
print(d1, ' ndim: ', d1.ndim, '\n') 
print(d2, ' ndim: ', d2.ndim, '\n') 
print(d3, ' ndim: ', d3.ndim, '\n') 
```

```python
### DataFrame 으로 부터 생성
import pandas as pd

mypd1=pd.DataFrame([1, 2, 3])
print(mypd1, '\n')

mynp1=np.array(mypd1)

print(mynp1, '\n'),  
print('shape = ', mynp1.shape)
```

```python
### pandas Series 로 부터 생성
pd1 = pd.Series([1,2,3])
np1 = np.array(pd1)
print(pd1, '\n')
print(np1)
print('shape =' , np1.shape, ' dim = ', np1.ndim)
```
