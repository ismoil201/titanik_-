```python
def normalize(df):
    """데이터를 0~1 사이 값으로 정규화합니다."""
    return (df - df.min()) / (df.max() - df.min())

# 정규 화(Min-Max Scaling) - 데이타를 0과 1사이의 값으로 압축.  기본 분포 그대로 유지 되도록 함.
#  y =   ( x - mean ) / (max - min)
titanic_df['age_norm'] = normalize(titanic_df['age'])
titanic_df
```

```python
def standardize(df):
    """평균 0, 표준편차 1이 되도록 표준화합니다."""
    return (df - df.mean()) / df.std()

# 표준화 (Z-score scaling) - 평균을 0, 표준편차를 1로 변환.   표준 정규 분포를 따르도록 유도.  데이타가 정규분포를 따를때
#  y = (x - mean) / 표준편차
titanic_df['age_std'] = standardize(titanic_df['age'])
titanic_df
```

###  분산 구하기:  var
## .var() :   sigma ** 2    분산
## .std() :  sigma   표준 편차   
## 
## 분산(variance) 과 표준편차(standard deviation) 는 각각 중심에서 얼마나 떨어져 있는지를 나타내는 지표
## variance:  원래 데이터의 제곱 단위  (x - mean) ** 2,  ==   sigma ** 2 .    수학적으로 다루기 쉬움. 머신 러닝 에서 미분 등..
## std:  원래 데이터의 단위 :   sqrt(variance) == sigma

## std: 중심값에서 떨어진 정도  
## 1sigma: 전체 데이터의 68% 정도가 mean 을 중심으로 그 안에 있슴 ,    2sigma: 중심값에서 95%,  3sigma:  중심값에서 99.7%  

## surviced std: 0.477275 의 해석
## 이진 데이터 (1, 0) 의 std 는 넓게 퍼진 정도가 아니라 데이터가 한쪽으로 쏠린 정도를 나타냄.   이 경우는 0.5 근처에 

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. 데이터 준비 (표준정규분포)
mu, sigma = 0, 1
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, mu, sigma)

plt.figure(figsize=(10, 6))
plt.plot(x, y, 'black', lw=2)

# 2. 각 구간 색칠하기 (중첩되므로 투명도 조절이 중요)
# 99.7% 영역 (3 sigma)
plt.fill_between(x, y, where=(x >= -3) & (x <= 3), color='lightgray', alpha=0.3, label='99.7% (3$\sigma$)')
# 95% 영역 (2 sigma)
plt.fill_between(x, y, where=(x >= -2) & (x <= 2), color='skyblue', alpha=0.4, label='95.0% (2$\sigma$)')
# 68% 영역 (1 sigma)
plt.fill_between(x, y, where=(x >= -1) & (x <= 1), color='royalblue', alpha=0.6, label='68.0% (1$\sigma$)')

# 3. 그래프 꾸미기
plt.title('Empirical Rule: 68-95-99.7', fontsize=15)
plt.xlabel('Standard Deviations ($\sigma$)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.xticks([-3, -2, -1, 0, 1, 2, 3])
plt.legend()
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
## 왜도 (skewness):   데이터가 어느 한쪽으로 치우쳤는가의 정도
##  0 에 가까움:  평균/중앙값/최빈값 이 거의 일치 함.  좌우 대칭
## + 값: 오른쪽으로 꼬리가 김
## - 값: 왼쪽으로 꼬리가 김

## 꼬리가 길게 치우쳐 있다는 뜻은 대부분의 사람들은 그 반대 편에 위치 한다는 뜻.   즉, 꼬리가 오른쪽으로 길면 왼쪽 부분이 다수 라는 뜻

## pclass:   3 등석에 몰려 있슴 (1 / 2 / 3 등석)
## fare:  요금이 오른쪽에 몰려 있슴 (비싼 요금)
titanic_df.skew(numeric_only=True )
```

```python
###  첨도(kurtosis):   뾰쪽한 정도.   얼마나 중앙에 몰려 있나.  표준편차에서는 0 이 중앙
##  첨도가 높을 수록 데이터가 평균 근처에 아주 빽빽하게 모여 있고, 동시에 아주 멀리 떨어진 극단적인 값(이상치)이 존재한다는 뜻입니다. (예: 500달러가 넘는 요금)
##  첨도가 낮을 수록 데이터가 중앙에 있지 않고 아주 넓게 퍼져 있슴

# survived 는 0 / 1  이라서 중앙이 비어 있슴.  낮은 첨도
# age:  그래도 0 보다 좀 커서 대체로 중앙값에 어느 정도 몰려 있슴
# parch:  아주 높은 밀도로 중앙에 몰려 있슴. (대부분 가족 없이 탑승)
# fare: 아주 멀리 떨어진 극단적인 값(이상치) (outliers)이 존재한다는 뜻입니다. (예: 500달러가 넘는 요금)

titanic_df.kurt(numeric_only=True)
```
