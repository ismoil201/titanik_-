```python
f=open('file.txt', 'rt', encoding='UTF8')
for line in f:
    print(line, end="")         # print 함수가 자동으로 '\n' 을 붙이는것을 방지
    #print(line)

f.close()
```

```python
## newline=''  :  읽어 올때 줄바꿈 문자를 임의로 변경하지 말고 그대로 읽어라
## windows: \r\n,  linux(mac): \n   임.   
## python 은  \r\n or \n  ==> \n 으로 변환해서 읽음(default 동작)  

with open("proverb.txt",'r', encoding='UTF8', newline='') as filereader:
  for row in filereader:
    print("{}".format(row.strip( )))      # strip 을 통해 양끝의 공백 및 \n 을 제거
    #print(row)

    
```

```python
# glob이라는 이름은 **Unix shell의 "globbing"**에서 왔다.
# glob module finds pathnames matching a specified pattern

import os
import glob     

for file in glob.glob(os.path.join('*.txt')):
  with open(file, 'r', newline='',encoding='UTF8') as filereader:
    for row in filereader:
      print("{}".format(row.strip( )))
```
